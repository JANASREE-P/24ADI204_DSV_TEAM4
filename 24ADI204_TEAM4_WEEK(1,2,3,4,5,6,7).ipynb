WEEK 7:DIMENISONALITY REDUCTION

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import numpy as np
```

Selecting input features for PCA

For dimensionality reduction, only numerical predictor columns are selected.
The target variable popularity is excluded from the input features because it is the value to be predicted.

```python
pca_features = ['danceability', 'energy', 'tempo', 'loudness',
                'duration_ms', 'key', 'mode', 'instrumentalness', 'stream_count']

X = df[pca_features]
y = df['popularity']

print("Input shape before PCA:", X.shape)
X.head()
```

Standardizing the features

The selected numerical features are standardized so that each feature has mean 0 and standard deviation 1.
This step is necessary because PCA works based on variance, and features with larger values can dominate the results if scaling is not done.

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Scaled data shape:", X_scaled.shape)
```

Applying PCA

PCA transforms the original features into a new set of variables called principal components.
These components are arranged in such a way that:

the first component captures the maximum variance,
the second captures the next highest variance,
and so on.

This helps reduce the number of features while retaining most of the dataset information.

```python
pca = PCA()
X_pca = pca.fit_transform(X_scaled)

explained_variance = pca.explained_variance_ratio_
cumulative_variance = np.cumsum(explained_variance)

print("Explained Variance Ratio:")
print(explained_variance)

print("\nCumulative Variance:")
print(cumulative_variance)
```

Visualizing explained variance

The explained variance plot shows how much information is captured by each principal component.
The cumulative variance curve helps decide how many components should be retained.

Usually, components explaining about 95% of total variance are selected.

```python
plt.figure(figsize=(8,5))
plt.plot(range(1, len(explained_variance) + 1), cumulative_variance, marker='o')
plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("Cumulative Explained Variance by PCA Components")
plt.grid(True)
plt.show()
```

Choosing optimal number of components

Here, PCA is applied again by keeping 95% of the total variance.
This automatically selects the minimum number of principal components needed to preserve most of the information.

```python
pca_final = PCA(n_components=0.95)
X_pca_final = pca_final.fit_transform(X_scaled)

print("Original shape:", X.shape)
print("Reduced shape after PCA:", X_pca_final.shape)
print("Number of components selected:", pca_final.n_components_)
```

PCA transformed dataset

The reduced dataset now contains fewer dimensions than the original feature set.
These principal components can be used as input for machine learning models instead of the original features.

This helps in:

reducing complexity,
speeding up model training,
and minimizing multicollinearity among features.

```python
pca_columns = [f'PC{i+1}' for i in range(X_pca_final.shape[1])]
df_pca = pd.DataFrame(X_pca_final, columns=pca_columns)

df_pca.head()
```

Feature contribution to principal components

The PCA component matrix shows how strongly each original feature contributes to the new principal components.
This helps in interpreting the transformed dimensions.

```python
loadings = pd.DataFrame(
    pca_final.components_.T,
    columns=pca_columns,
    index=pca_features
)

loadings
```

Model training using PCA features

After dimensionality reduction, the transformed PCA features are used to train the regression model.
This allows us to compare model performance before and after dimensionality reduction.

```python
from xgboost import XGBRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

X_train, X_test, y_train, y_test = train_test_split(
    X_pca_final, y, test_size=0.2, random_state=42
)

model_pca = XGBRegressor(
    n_estimators=500,
    learning_rate=0.05,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
    n_jobs=-1,
    random_state=42
)

model_pca.fit(X_train, y_train)

y_pred_pca = model_pca.predict(X_test)

print("PCA Model MSE:", mean_squared_error(y_test, y_pred_pca))
print("PCA Model R2 Score:", r2_score(y_test, y_pred_pca))
```

Conclusion

PCA successfully reduced the dimensionality of the dataset by converting the original numerical features into a smaller number of principal components while preserving most of the variance.

This step is useful because it:

simplifies the feature space,
reduces redundancy,
and supports efficient model building.

The reduced PCA features can now be used for further predictive modeling and performance comparison.
